# 🐾 Project Description
In this capstone project for the Google/Kaggle Gen AI Intensive Course 2025Q1, we present **PetCare AI Assistant** — a GenAI-powered solution designed to support pet owners by providing intelligent, personalized advice on common pet health, grooming, and dietary concerns.

The assistant analyzes real-world veterinary data sourced from VetDataHub and uses **Google Gemini (Gemini Pro)** to:
- Understand pet symptoms or medical notes
- Provide a preliminary diagnosis or action plan
- Recommend relevant services (e.g., vet consultation, grooming, blood test)
- Output structured responses in JSON format for downstream automation

# **Step 1: Install & Import Dependencies**

````python
!pip install -q sentence-transformers PyMuPDF
import os
import json
import numpy as np
import pandas as pd
from IPython.display import display, Markdown
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import google.generativeai as genai
import fitz  # PyMuPDF
import ipywidgets as widgets
````

# **Step 2: Configure Gemini API**

````python
from kaggle_secrets import UserSecretsClient
user_secrets = UserSecretsClient()
GOOGLE_API_KEY = user_secrets.get_secret("GOOGLE_API_KEY")
# Configuring the Gemini API
genai.configure(api_key=GOOGLE_API_KEY)
model = genai.GenerativeModel('gemini-2.0-flash')
````

# 📂 Step 3: Load Veterinary Documents

````python
def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    text = ""
    for page in doc:
        text += page.get_text()
    return text

pdf1 = extract_text_from_pdf("/kaggle/input/vet-dataset/pdf1.pdf")
pdf2 = extract_text_from_pdf("/kaggle/input/vet-dataset/pdf2.pdf")
documents = [pdf1, pdf2]
````

# 📄 Step 4: Chunk Documents into Paragraphs

````python
"""
RAG systems work by chunking large documents into manageable parts for embedding and retrieval.
"""

chunks = []
for idx, doc in enumerate(documents):
    for para in doc.split(". "):
        if len(para.strip()) > 50:
            chunks.append({"doc_id": f"doc_{idx+1}", "text": para.strip()})
chunk_df = pd.DataFrame(chunks)

embed_model = SentenceTransformer('all-MiniLM-L6-v2')
chunk_df["embedding"] = chunk_df["text"].apply(lambda x: embed_model.encode(x, convert_to_numpy=True))
````

# ✍️ Step 5: Define Prompt Template and Get Gemini Response

````python
few_shot_example = """
Example:
Pet Type: Cat
Symptoms: vomiting, hairball, no appetite
Context: Hairballs in cats can cause vomiting and appetite loss.

Output:
{
  "problem_category": "health",
  "preliminary_diagnosis": "Hairball-induced gastrointestinal upset",
  "suggested_action": "Provide hydration, anti-hairball paste, vet visit if persistent",
  "recommended_services": ["Vet Consultation"],
  "urgency": "Medium"
}
"""

def get_response(query, pet_type="Dog"):
    q_embed = embed_model.encode(query, convert_to_numpy=True).reshape(1, -1)
    chunk_matrix = np.vstack(chunk_df["embedding"])
    sims = cosine_similarity(q_embed, chunk_matrix)[0]
    top_idxs = sims.argsort()[-3:][::-1]
    context = "\n".join(chunk_df.iloc[i]["text"] for i in top_idxs)

    prompt = f"""
{few_shot_example}

Pet Type: {pet_type}
Symptoms: {query}
Context: {context}

Please return only raw JSON without code block formatting.
"""
    response = model.generate_content(prompt)
    return response.text
````

# ✅ Step 6: Parse Response

````python
def clean_and_parse_json(raw_text):
    try:
        cleaned = raw_text.strip()
        # Remove any markdown code fences
        if cleaned.startswith("```json"):
            cleaned = cleaned[len("```json"):].strip()
        if cleaned.endswith("```"):
            cleaned = cleaned[:-3].strip()
        return json.loads(cleaned)
    except Exception as e:
        return {
            "error": str(e),
            "raw_response": raw_text
        }
````

# 🧪 Step 7: Evaluate Against Known Diagnosis

````python
def evaluate_output(gemini_text, actual_text):
    e1 = embed_model.encode(gemini_text)
    e2 = embed_model.encode(actual_text)
    score = cosine_similarity([e1], [e2])[0][0]
    return round(score, 2)

# 💬 Step 10: Interactive UI for Query
query_box = widgets.Textarea(
    value="My dog is vomiting and seems weak.",
    placeholder='Describe symptoms here...',
    description='Symptoms:',
    layout=widgets.Layout(width='90%', height='100px')
)
run_button = widgets.Button(description="Analyze", button_style='success')
output_box = widgets.Output()
````

````python
def on_run_clicked(_):
    with output_box:
        output_box.clear_output()
        q = query_box.value
        print("🔍 User Query:", q)
        raw = get_response(q)
        result = clean_and_parse_json(raw)
        df = pd.DataFrame([result])
        styled = df.style.set_properties(**{'white-space': 'pre-wrap'})
        display(styled)
        # Optional evaluation with known answer
        known = "Recommend hydration therapy, switch to bland diet, administer antiemetic if vomiting continues."
        predicted = result.get("suggested_action", "")
        score = evaluate_output(predicted, known)
        print("🧪 Similarity Score with Vet Plan:", score)

run_button.on_click(on_run_clicked)
display(query_box, run_button, output_box)
````
